```python

import numpy as np
import matplotlib.pyplot as plt

# Define the function
def f(x):
    return 1 / x**2

# Create x values: avoid 0 by splitting into two domains
x1 = np.linspace(-3, -0.05, 500)   # left branch
x2 = np.linspace(0.05, 3, 500)     # right branch

y1 = f(x1)
y2 = f(x2)

# Plot setup
plt.figure(figsize=(6, 6))
plt.plot(x1, y1, 'b-', linewidth=2, label=r'$y = 1/x^2$')
plt.plot(x2, y2, 'b-', linewidth=2)

# Asymptotes (dashed lines)
plt.axhline(0, color='gray', linestyle='--', linewidth=1)   # horizontal asymptote
plt.axvline(0, color='gray', linestyle='--', linewidth=1)   # vertical asymptote

# Labels and limits
plt.xlim(-3, 3)
plt.ylim(0, 10)        # y from 0 to 10 (since function →∞ at 0)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Graph of $y = 1/x^2$')
plt.legend()
plt.grid(True, alpha=0.3)

# Show the plot
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return 1 / x

# Split domain to avoid x=0
x1 = np.linspace(-3, -0.05, 500)   # negative side
x2 = np.linspace(0.05, 3, 500)     # positive side

y1 = f(x1)
y2 = f(x2)

plt.figure(figsize=(6, 6))
plt.plot(x1, y1, 'r-', linewidth=2, label=r'$y = 1/x$')
plt.plot(x2, y2, 'r-', linewidth=2)

# Asymptotes
plt.axhline(0, color='gray', linestyle='--', linewidth=1)
plt.axvline(0, color='gray', linestyle='--', linewidth=1)

plt.xlim(-3, 3)
plt.ylim(-3, 3)   # symmetric range
plt.xlabel('x')
plt.ylabel('y')
plt.title('Graph of $y = 1/x$')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2 / (x - 3)

# Create x values, avoiding x=3
x1 = np.linspace(-5, 2.9, 500)   # left branch
x2 = np.linspace(3.1, 10, 500)   # right branch

y1 = f(x1)
y2 = f(x2)

plt.figure(figsize=(8, 6))

# Plot curve
plt.plot(x1, y1, 'b-', linewidth=2, label=r'$f(x) = x^2/(x-3)$')
plt.plot(x2, y2, 'b-', linewidth=2)

# Asymptotes
plt.axvline(3, color='red', linestyle='--', linewidth=1, label='Vertical asymptote x=3')
# Slant asymptote y = x+3
x_line = np.linspace(-5, 10, 100)
plt.plot(x_line, x_line + 3, 'gray', linestyle=':', linewidth=1, label='Slant asymptote y = x+3')

# Axes and limits
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.xlim(-5, 10)
plt.ylim(-20, 30)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Graph of $f(x) = \\frac{x^2}{x-3}$')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import sympy as sp
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# ------------------
# Symbolic analysis
# ------------------
x = sp.symbols('x')
y = sp.exp(1/x)

# First derivative
y_prime = sp.diff(y, x)

# Second derivative
y_double_prime = sp.diff(y_prime, x)

print("y =", y)
print("y' =", sp.simplify(y_prime))
print("y'' =", sp.simplify(y_double_prime))

# Limits
print("\nLimits:")
print("x -> 0+ :", sp.limit(y, x, 0, '+'))
print("x -> 0- :", sp.limit(y, x, 0, '-'))
print("x -> ∞  :", sp.limit(y, x, sp.oo))
print("x -> -∞ :", sp.limit(y, x, -sp.oo))

# ------------------
# Numerical plotting
# ------------------

# Convert symbolic expression to numerical function
f = sp.lambdify(x, y, modules=['numpy'])

# Avoid x = 0
x_left = np.linspace(-5, -0.05, 1000)
x_right = np.linspace(0.05, 5, 1000)

y_left = f(x_left)
y_right = f(x_right)

# Seaborn style
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))

sns.lineplot(x=x_left, y=y_left, label=r"$e^{1/x}$")
sns.lineplot(x=x_right, y=y_right)

# Horizontal asymptote y = 1
plt.axhline(1, linestyle="--", label="y = 1")

# Vertical asymptote x = 0
plt.axvline(0, linestyle="--", label="x = 0")

# Some notable points
plt.scatter([-1, -0.5, 1],
            [np.exp(-1), np.exp(-2), np.exp(1)])

plt.title(r"Graph of $y=e^{1/x}$")
plt.xlabel("x")
plt.ylabel("y")
plt.ylim(-0.2, 10)
plt.legend()
plt.show()
```

```python
import sympy as sp
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Symbolic function
x = sp.symbols('x')
y = x * sp.exp(x)

# Convert to numerical function
f = sp.lambdify(x, y, 'numpy')

# Generate x values
x_vals = np.linspace(-5, 3, 1000)
y_vals = f(x_vals)

# Important points
min_x = -1
min_y = f(min_x)

infl_x = -2
infl_y = f(infl_x)

# Seaborn styling
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))

# Main curve
sns.lineplot(x=x_vals, y=y_vals, linewidth=2,
             label=r"$y=xe^x$")

# Axes
plt.axhline(0, linestyle="--", alpha=0.7)
plt.axvline(0, linestyle="--", alpha=0.7)

# Minimum point
plt.scatter(min_x, min_y, s=80,
            label=f"Minimum (-1, {-1/np.e:.3f})")

# Inflection point
plt.scatter(infl_x, infl_y, s=80,
            label=f"Inflection (-2, {-2/np.e**2:.3f})")

# Labels
plt.title(r"Graph of $y = xe^x$")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()

plt.show()
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sympy import symbols, Eq
from sympy import init_printing
from IPython.display import display

# ==========================================
# 1. Define the equation using SymPy
# ==========================================

init_printing()  # Pretty mathematical display in Jupyter

x = symbols('x')
y_expr = x**2 - 5*x + 6

equation = Eq(symbols('y'), y_expr)

print("Quadratic Equation:")
display(equation)

# ==========================================
# 2. Create the data table
# ==========================================

x_values = np.arange(-1, 7)

y_values = [y_expr.subs(x, val) for val in x_values]

df = pd.DataFrame({
    'x': x_values,
    'y = x² - 5x + 6': y_values
})

print("\nTable of Values:")
display(df)

# ==========================================
# 3. Calculate important features
# ==========================================

a = 1
b = -5
c = 6

# Axis of symmetry
axis_x = -b / (2 * a)

# Vertex
vertex_y = y_expr.subs(x, axis_x)

print(f"Axis of Symmetry: x = {axis_x}")
print(f"Vertex: ({axis_x}, {vertex_y})")

# ==========================================
# 4. Plot the parabola
# ==========================================

x_plot = np.linspace(-2, 7, 500)
y_plot = x_plot**2 - 5*x_plot + 6

plt.figure(figsize=(10, 6))

# Parabola
plt.plot(x_plot, y_plot, linewidth=2,
         label=r'$y=x^2-5x+6$')

# Data points from the table
plt.scatter(x_values, y_values, s=80,
            label='Table Points')

# Axis of symmetry
plt.axvline(axis_x,
            linestyle='--',
            linewidth=2,
            label=f'Axis of Symmetry x={axis_x}')

# Vertex
plt.scatter(axis_x, float(vertex_y),
            s=120,
            marker='*',
            label=f'Vertex ({axis_x}, {vertex_y})')

# Annotate points
for xv, yv in zip(x_values, y_values):
    plt.annotate(
        f'({xv},{yv})',
        (xv, yv),
        xytext=(5, 5),
        textcoords='offset points'
    )

# Annotate vertex
plt.annotate(
    f'Vertex\n({axis_x},{vertex_y})',
    (axis_x, float(vertex_y)),
    xytext=(10, -30),
    textcoords='offset points'
)

plt.title('Graph of $y=x^2-5x+6$')
plt.xlabel('x')
plt.ylabel('y')

plt.grid(True)
plt.legend()

plt.xlim(-2, 7)
plt.ylim(-4, 14)

plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return 1 / (x**2 - 1)

# x values avoiding discontinuities at ±1
x1 = np.linspace(-5, -1.01, 400)
x2 = np.linspace(-0.99, 0.99, 400)
x3 = np.linspace(1.01, 5, 400)

y1 = f(x1)
y2 = f(x2)
y3 = f(x3)

plt.figure(figsize=(10, 6))
plt.plot(x1, y1, 'b-', label=r'\( f(x) = \frac{1}{x^2 - 1} \)')
plt.plot(x2, y2, 'b-')
plt.plot(x3, y3, 'b-')

plt.axvline(x=-1, color='r', linestyle='--', label='Vertical asymptotes')
plt.axvline(x=1, color='r', linestyle='--')
plt.axhline(y=0, color='g', linestyle='--', label='Horizontal asymptote y=0')

plt.title(r'Sketch of \( f(x) = \frac{1}{x^2 - 1} \)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.ylim(-10, 10)
plt.xlim(-5, 5)

#plt.savefig('function_plot.png')
plt.show()  # This would display it in a local environment
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Define the function
def f(x):
    return 1 / (x**2 - 1)

# Create x values, avoiding the vertical asymptotes at -1 and 1
x = np.linspace(-3, 3, 400)
x = x[(x != -1) & (x != 1)]  # remove exact asymptote points

# Compute y values
y = f(x)

# Plot the function
plt.figure(figsize=(8,6))
plt.plot(x, y, label=r"$f(x) = \frac{1}{x^2 - 1}$", color="blue")

# Add vertical asymptotes
plt.axvline(x=-1, color="red", linestyle="--", label="x = -1")
plt.axvline(x=1, color="red", linestyle="--", label="x = 1")

# Add horizontal asymptote
plt.axhline(y=0, color="green", linestyle="--", label="y = 0")

# Add intercept point
plt.scatter(0, f(0), color="black", zorder=5)

# Labels and legend
plt.title("Graph of f(x) = 1 / (x^2 - 1)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.ylim(-10, 10)  # limit y for better view
plt.legend()
plt.grid(True)
plt.show()
```

```python

from sympy import *
from IPython.display import display, Math
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# SymPy Analysis
# =====================================================

x = symbols('x', real=True)

f = exp(x)
g = log(x)

display(Math(r"\textbf{Functions}"))
display(Eq(Symbol('y'), f))
display(Eq(Symbol('y'), g))

# Derivatives
display(Math(r"\textbf{First Derivatives}"))
display(Eq(Symbol("d_dx_e"), diff(f, x)))
display(Eq(Symbol("d_dx_ln"), diff(g, x)))

# Second derivatives
display(Math(r"\textbf{Second Derivatives}"))
display(Eq(Symbol("d2_dx2_e"), diff(f, x, 2)))
display(Eq(Symbol("d2_dx2_ln"), diff(g, x, 2)))

# Limits
display(Math(r"\textbf{Limits}"))

display(
    Eq(
        Limit(exp(x), x, -oo),
        limit(exp(x), x, -oo)
    )
)

display(
    Eq(
        Limit(exp(x), x, oo),
        limit(exp(x), x, oo)
    )
)

display(
    Eq(
        Limit(log(x), x, 0, dir='+'),
        limit(log(x), x, 0, dir='+')
    )
)

display(
    Eq(
        Limit(log(x), x, oo),
        limit(log(x), x, oo)
    )
)

# =====================================================
# Pandas Tables
# =====================================================

exp_values = [-2, -1, 0, 1, 2]

df_exp = pd.DataFrame({
    "x": exp_values,
    "e^x": [float(N(exp(v), 6)) for v in exp_values]
})

ln_values = [0.25, 0.5, 1, 2, 4]

df_ln = pd.DataFrame({
    "x": ln_values,
    "ln(x)": [float(N(log(v), 6)) for v in ln_values]
})

print("\nTable of e^x")
print(df_exp)

print("\nTable of ln(x)")
print(df_ln)

# =====================================================
# Matplotlib Sketch
# =====================================================

x_exp = np.linspace(-3, 3, 1000)
y_exp = np.exp(x_exp)

x_ln = np.linspace(0.01, 5, 1000)
y_ln = np.log(x_ln)

plt.figure(figsize=(12, 7))

# Curves
plt.plot(x_exp, y_exp, linewidth=2,
         label=r'$y=e^x$')

plt.plot(x_ln, y_ln, linewidth=2,
         label=r'$y=\ln(x)$')

# Reflection line
plt.plot(
    [-3, 5],
    [-3, 5],
    linestyle='--',
    label=r'$y=x$'
)

# Axes
plt.axhline(0)
plt.axvline(0)

# Asymptotes
plt.axhline(
    0,
    linestyle=':',
    linewidth=1,
    label=r'Asymptote $y=0$'
)

plt.axvline(
    0,
    linestyle=':',
    linewidth=1,
    label=r'Asymptote $x=0$'
)

# Important points
plt.scatter([0], [1], s=60)
plt.annotate(
    "(0,1)",
    xy=(0, 1),
    xytext=(0.3, 1.3)
)

plt.scatter([1], [0], s=60)
plt.annotate(
    "(1,0)",
    xy=(1, 0),
    xytext=(1.2, 0.3)
)

# Labels
plt.title(r"Sketch of $e^x$ and $\ln(x)$")
plt.xlabel("x")
plt.ylabel("y")

plt.xlim(-3, 5)
plt.ylim(-3, 10)

plt.grid(True)
plt.legend()

plt.show()
```
