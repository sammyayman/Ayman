```python
import matplotlib.pyplot as plt
import numpy as np

# Create some sample data
x = np.linspace(0, 10, 100)

# Create figure and 2x3 subplots
fig, axes = plt.subplots(2, 3, figsize=(12, 6))

# First subplot
axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title("Sine")

# Second subplot
axes[0, 1].plot(x, np.cos(x))
axes[0, 1].set_title("Cosine")

# Third subplot
axes[0, 2].plot(x, np.tan(x))
axes[0, 2].set_title("Tangent")

# Fourth subplot
axes[1, 0].scatter(x, np.sin(x))
axes[1, 0].set_title("Scatter")

# Fifth subplot
axes[1, 1].bar([1,2,3,4], [5,7,3,6])
axes[1, 1].set_title("Bar Chart")

# Sixth subplot
axes[1, 2].hist(np.random.randn(1000), bins=20)
axes[1, 2].set_title("Histogram")

# Adjust spacing
plt.tight_layout()

# Show all diagrams
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample x values
x = np.linspace(0, 10, 100)

# Create 2x3 subplot
fig, axes = plt.subplots(2, 3, figsize=(12, 6))

# Flatten axes array to make looping easier
axes = axes.flatten()

# Create 6 diagrams using for loop
for i in range(6):

    # Example: different sine waves
    y = np.sin(x + i)

    # Plot on each subplot
    axes[i].plot(x, y)

    # Title
    axes[i].set_title(f"Diagram {i+1}")

# Adjust spacing
plt.tight_layout()

# Show all plots
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)

plt.figure(figsize=(8, 5))

for n in range(1, 5):
    y = np.sin(n * x) / n
    plt.plot(x, y, label=f'sin({n}x)/{n}')

plt.title("Sine waves with a for loop")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()






data = [np.random.normal(0, std, 200) for std in [1, 2, 3, 4]]

fig, axes = plt.subplots(2, 2, figsize=(8, 6))

for ax, std, d in zip(axes.flatten(), [1, 2, 3, 4], data):
    ax.hist(d, bins=20, alpha=0.7)
    ax.set_title(f'Std = {std}')
    ax.set_xlabel('Value')

plt.tight_layout()
plt.show()










colors = ['red', 'blue', 'green']
A = np.random.rand(5, 2)
B = A+1
C = B+1
groups = {'A': A,
          'B': B,
          'C': C}
print(B)

plt.figure(figsize=(6, 6))

for i, (label, points) in enumerate(groups.items()):
    plt.scatter(points[:, 0], points[:, 1], c=colors[i], label=label, s=80)

plt.legend()
plt.grid()
plt.title("Scatter plot with loop")
plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np

# Create some data
x = np.linspace(0, 10, 100)
y_values = [np.sin(x), np.cos(x), np.tan(x)]
titles = ['Sine Wave', 'Cosine Wave', 'Tangent Wave']
colors = ['r', 'g', 'b']

# Create a figure with 2 rows and 2 columns of subplots
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Plot sine, cosine, and tangent using a for loop
for i in range(3):
    axs[i // 2, i % 2].plot(x, y_values[i], colors[i])  # Use integer division and modulo for indexing
    axs[i // 2, i % 2].set_title(titles[i])
    axs[i // 2, i % 2].set_xlabel('X-axis')
    axs[i // 2, i % 2].set_ylabel('Y-axis')

    if i == 2:  # Set limits for tangent plot
        axs[i // 2, i % 2].set_ylim(-10, 10)

# Fourth subplot (bottom-right) for histogram
axs[1, 1].hist(np.random.randn(1000), bins=30, color='purple', alpha=0.7)
axs[1, 1].set_title('Histogram of Random Data')
axs[1, 1].set_xlabel('Bins')
axs[1, 1].set_ylabel('Frequency')

# Adjust layout
plt.tight_layout()

# Show the plot
plt.show()
```

```python

import matplotlib.pyplot as plt
#import numpy as np

fig, axes = plt.subplots(2, 2)
a = np.arange(0, 10, 0.5)

# Store the unique plot data for A, B, and C
# Format: ( [line1_args], [line2_args], ... )
plot_data = [
    ([ [10,20,30,40] ], [ [1,2,3,4] ]),

    ([ [10,20,30,40], [1,2,3,4], 'r*-' ],
[ [5,10,15,20], [1,2,3,4], 'c*-' ],
[ [2,4,6,8], [1,2,3,4], 'k.-' ],
 [ [11,22,33,42], [1,2,3,4], 'b.-' ]),

    ([ [10,20,30,40], [1,2,3,4], 'k^-.' ],
[ [5,10,15,20], [1,2,3,4], '--' ],
 [ [2,4,6,8], [1,2,3,4], 'o:' ])
]

# Loop through all 4 subplots dynamically
for i, ax in enumerate(axes.flat):
    letter = ['A', 'B', 'C', 'D'][i]

    if letter == 'D':
        ax.plot(a, 'go:', a*3, 'r>--', a**2, 'c+-')
    else:
        for line in plot_data[i]:
            ax.plot(*line) # * unpacks the list arguments into ax.plot()

    # Standard formatting for all plots
    ax.set_xlabel('X Label')
    ax.set_ylabel('Y Label')
    ax.set_title(f'Plot {letter}')

plt.tight_layout()
plt.show()
del a ,np
```

```python


# Example 1: Multiple line plots
print("Example 1: Multiple line plots")
plt.figure(figsize=(10, 5))
for i in range(1, 4):
    y = [x * i for x in range(10)]
    plt.plot(y, label=f'y = x * {i}')
plt.legend()
plt.title('Multiple Lines')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()


# Example 2: Subplots in a grid
print("Example 2: Subplots in a grid")
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
for i, ax in enumerate(axes.flat):
    data = [j ** (i + 1) for j in range(1, 6)]
    ax.bar(range(len(data)), data, color='steelblue')
    ax.set_title(f'Power of {i + 1}')
plt.tight_layout()
plt.show()


# Example 3: Multiple scatter plots with different colors
print("Example 3: Multiple scatter plots")
plt.figure(figsize=(10, 5))
colors = ['red', 'blue', 'green', 'orange']
for i, color in enumerate(colors):
    x = [j for j in range(10)]
    y = [val + i * 5 for val in range(10)]
    plt.scatter(x, y, color=color, label=f'Dataset {i + 1}', s=50)
plt.legend()
plt.title('Multiple Scatter Plots')
plt.show()


# Example 4: Drawing multiple shapes (rectangles)
print("Example 4: Multiple rectangles")
fig, ax = plt.subplots(figsize=(8, 8))
for i in range(5):
    width = 1 + i * 0.4
    height = 1 + i * 0.4
    rect = plt.Rectangle((i, i), width, height,
                          fill=False, edgecolor='purple', linewidth=2)
    ax.add_patch(rect)


ax.set_xlim(-1, 8)
ax.set_ylim(-1, 8)
ax.set_aspect('equal')
plt.title('Multiple Rectangles')
plt.show()


# Example 5: Histogram with multiple datasets
print("Example 5: Multiple histograms")
import numpy as np
plt.figure(figsize=(10, 5))
data_sets = [
    np.random.normal(0, 1, 100),
    np.random.normal(2, 1, 100),
    np.random.normal(4, 1, 100)
]
for i, data in enumerate(data_sets):
    plt.hist(data, bins=20, alpha=0.6, label=f'Dataset {i + 1}')
plt.legend()
plt.title('Multiple Histograms')
plt.show()


# Example 6: Annotating multiple points
print("Example 6: Annotating multiple points")
plt.figure(figsize=(10, 5))
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
plt.plot(x, y, 'bo-')
for i in range(len(x)):
    plt.annotate(f'({x[i]}, {y[i]})',
                xy=(x[i], y[i]),
                xytext=(5, 5),
                textcoords='offset points')
plt.title('Annotated Points')
plt.show()
```
